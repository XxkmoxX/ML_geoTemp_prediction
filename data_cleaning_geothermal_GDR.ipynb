```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import csv

df = pd.read_csv('csv/dataset_part1.csv')

print('\nDataFrame shape: ', df.shape)
print('\nDataFrame info: ', df.info())

#print('\nColumn headers:')
#for i, col in enumerate(df):
#    print(f'{i+1}, {col}')

print('\nFirst 10 rows: ', df.head(10))
print('\nLast 10 rows: ', df.tail(10))

df.columns.to_list()
```

```python
### Princial characteristics of dataset

print('\nColumn names:\n', df.columns.tolist())
print('\nData types:\n', df.dtypes)
print('\nMissing values:\n', df.isnull().sum())
print('\nNaN entries:\n', df.isna().sum())
print('\nDuplicates Count:', df.duplicated().sum())

print('\nSummary Statistics:\n', df.describe(include='all').T)
```

```python
### Eliminate rows that are not wells

original_len = len(df)

# Create a pattern to match using regex
pattern = r'spring|fumarole|thermal spring'

df = df[~df['well_sample'].str.lower().str.contains(pattern, na=False)]

print(f'Original number of rows: {original_len}')
print(f'Number of rows after filtering: {len(df)}')
print(f'Number of rows removed: {original_len - len(df)}')

print('\nFirst 10 rows of cleaned dataset:')
display(df.head(10))
```

```python
### Keep only rows that contain "well" or "Well"

original_len = len(df)

# Create a pattern to match "well"
pattern = r'well'

df = df[df['well_sample'].str.lower().str.contains(pattern, na=False)]

print(f'Original number of rows: {original_len}')
print(f'Number of rows after filtering: {len(df)}')
print(f'Number of rows removed: {original_len - len(df)}')

print('\nFirst 10 rows of cleaned dataset:')
display(df.head(10))
```

```python
### Change commas for dots

numeric_cols = df.select_dtypes(include=['object']).columns

for col in numeric_cols:
    df[col] = df[col].astype(str).str.replace(',', '.')
   
    try:
        df[col] = pd.to_numeric(df[col])
    except:
        pass  # Keep as string if conversion fails
    
print("\nFirst 10 rows after replacing commas with dots:")
display(df.head(10))

print("\nUpdated data types:")
print(df.dtypes)
```

```python
### Check for NaN and missing values. Eliminate rows with more than 0 NaN

columns_to_check = ['pH', 'Na', 'K', 'Ca', 'Mg', 'Cl', 'SO4']

# First, replace string 'nan' with numpy.nan
df[columns_to_check] = df[columns_to_check].replace('nan', np.nan) # 'nan' not a number in Pandas. Numpy takes both.
df[columns_to_check] = df[columns_to_check].replace('NaN', np.nan) # 'NaN' not a number in Python.

# Count missing values
nan_counts = df[columns_to_check].isna().sum()
print("Missing value counts in each column:")
print(nan_counts)

# Print unique values to check for other potential missing value representations
for col in columns_to_check:
    print(f"\nUnique values in {col}:")
    print(df[col].unique())

# Count how many missing values each row has
row_nan_counts = df[columns_to_check].isna().sum(axis=1)

# Print total rows before filtering
print(f"\nTotal rows before filtering: {len(df)}")

# Keep only rows with 1 or fewer NaN values in the specified columns
df_cleaned = df[row_nan_counts <= 1]

print(f"Total rows after filtering: {len(df_cleaned)}")
print(f"Rows removed: {len(df) - len(df_cleaned)}")

df = df_cleaned

print("\nFirst few rows of cleaned dataset:")
display(df.head(10))
display(df.tail(10))
```

```python
### Eliminate wells with temperature ≤ 150°C

original_len = len(df)

# Filter to keep only wells with temperature > 150
df = df[df['temp_measured'] <= 180]

print(f'Original number of rows: {original_len}')
print(f'Number of rows after filtering: {len(df)}')
print(f'Number of rows removed: {original_len - len(df)}')

print('\nSummary statistics for temperature after filtering:')
print(df['temp_measured'].describe())

print('\nFirst 5 rows of filtered dataset:')
display(df.head())
display(df.tail())
```

```python
### Display groups of samples with the same name

# Get counts of each unique name
name_counts = df['well_sample'].value_counts()

# Filter for names that appear more than once
duplicate_names = name_counts[name_counts > 1]

print(f"Number of names that appear multiple times: {len(duplicate_names)}")
print("\nNames and their occurrence counts:")
print(duplicate_names)

print("\nDetailed view of each group:")
for name in duplicate_names.index:
    print(f"\n{'='*50}")
    print(f"Group: {name} (appears {name_counts[name]} times)")
    print(f"{'='*50}")
    display(df[df['well_sample'] == name])
```

```python
### Eliminate all samples with duplicate well names

original_len = len(df)

name_counts = df['well_sample'].value_counts()

# Filter for names that appear more than once
duplicate_names = name_counts[name_counts > 1]

print(f"Number of names that appear multiple times: {len(duplicate_names)}")
print("\nNames and their occurrence counts:")
print(duplicate_names)

# Remove all rows where well_sample appears more than once
df = df[~df['well_sample'].isin(duplicate_names.index)]

print(f'\nOriginal number of rows: {original_len}')
print(f'Number of rows after removing duplicates: {len(df)}')
print(f'Number of rows removed: {original_len - len(df)}')

# Verify no duplicates remain
remaining_duplicates = df['well_sample'].value_counts()
remaining_duplicates = remaining_duplicates[remaining_duplicates > 1]
print(f'\nRemaining duplicate well names: {len(remaining_duplicates)}')

print('\nFirst 5 rows of cleaned dataset:')
display(df.head())
display(df.tail())
```

```python
### Change str, e.g. '< 0.01' for numeric value (0.01)

# Define the columns to check
#columns_to_check = ['pH', 'K', 'Na', 'Ca', 'Mg', 'Cl', 'SO4']

columns_to_check = ['pH', 'K', 'Na', 'Ca', 'Mg', 'Cl', 'SO4']

check = '< .07'
value = 0.07

# Before replacement: Check for the presence of the str
print("Before replacement:")
for col in columns_to_check:
    # Count occurrences of the string
    count_before = df[df[col] == check].shape[0]
    print(f"Number of {check} values in {col}: {count_before}")
    # Show some example rows containing the value
    print(f"Example rows with {check} in {col}:")
    print(df[df[col] == check].head())

# Perform the replacement
for col in columns_to_check:
    df[col] = df[col].replace(check, value)
    #df[col] = pd.to_numeric(df[col], errors='coerce') # this function change all str to NaN, caution!!

# After replacement: Verify the changes
print("\nAfter replacement:")
for col in columns_to_check:
    # Count occurrences 
    count_after = df[df[col] == value].shape[0]
    print(f"\nNumber of {value} values in {col}: {count_after}")
    print(f"Unique values in {col}:")
    print((df[col].unique()))
```

```python
### Check for non-numeric 

columns_to_check = ['pH', 'K', 'Na', 'Ca', 'Mg', 'Cl', 'SO4']

print("Checking for non-numeric values:")
for col in columns_to_check:
    # Check if column is numeric
    is_numeric = pd.to_numeric(df[col], errors='coerce')
    
    # Find non-numeric values (will be NaN after coercion)
    non_numeric_mask = is_numeric.isna() & ~df[col].isna()
    non_numeric_values = df[non_numeric_mask][col]
    
    if len(non_numeric_values) > 0:
        print(f"\nColumn {col} has {len(non_numeric_values)} non-numeric values:")
        print("Sample of non-numeric values:")
        print(non_numeric_values.head())
        print("Unique non-numeric values:", df[non_numeric_mask][col].unique())
    else:
        print(f"\nColumn {col}: All values are numeric")
        print(f"Data type: {df[col].dtype}")
```
